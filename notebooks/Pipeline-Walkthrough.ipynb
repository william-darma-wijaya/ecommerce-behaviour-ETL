One row per day, no gaps: `days` should equal the number of days between first and last.

## dim_item (SCD Type 2)
How many versions do items have, and how many are placeholders?

```python
con.sql("""
    SELECT count(*) AS versions, count(DISTINCT item_id) AS items,
           count(*) FILTER (WHERE is_inferred)       AS placeholders,
           count(*) FILTER (WHERE NOT is_complete)   AS incomplete,
           count(*) FILTER (WHERE category_key = '-1') AS unknown_category
    FROM gold.dim_item
""").df()
```

```python
con.sql("""
    SELECT versions_per_item, count(*) AS items
    FROM (SELECT item_id, count(*) AS versions_per_item FROM gold.dim_item GROUP BY 1)
    GROUP BY 1 ORDER BY 1
""").df()
```

One item with several versions: each `valid_to` should equal the next `valid_from`.

```python
con.sql("""
    SELECT * FROM gold.dim_item
    WHERE item_id = (SELECT item_id FROM gold.dim_item GROUP BY 1 ORDER BY count(*) DESC LIMIT 1)
    ORDER BY valid_from
""").df()
```

# After Day 1 (2015060100)
`dbt run --vars "{batch_id: 2015060100}"`: silver and gold for the first day with events.

## Silver: bronze = clean + quarantine
Every bronze row of the batch ends up in exactly one of the two tables.

```python
con.sql("""
    WITH counts AS (
        SELECT
            (SELECT count(*) FROM bronze.brz_events            WHERE batch_id = '2015060100') AS bronze,
            (SELECT count(*) FROM silver.slv_events            WHERE batch_id = '2015060100') AS clean,
            (SELECT count(*) FROM silver.slv_events_quarantine WHERE batch_id = '2015060100') AS quarantine
    )
    SELECT *, bronze - clean - quarantine AS difference FROM counts
""").df()
```

Why were rows quarantined?

```python
con.sql("""
    SELECT reasons, count(*) AS row_count
    FROM silver.slv_events_quarantine WHERE batch_id = '2015060100'
    GROUP BY 1 ORDER BY 2 DESC
""").df()
```

A duplicate next to its original: the copy in quarantine, the first one in `slv_events`.

```python
con.sql("""
    SELECT timestamp, visitorid, event, itemid, transactionid, source_row_number, reasons
    FROM silver.slv_events_quarantine
    WHERE batch_id = '2015060100' AND reasons = 'DUPLICATE'
    LIMIT 5
""").df()
```

```python
con.sql("SELECT * FROM silver.slv_events WHERE batch_id = '2015060100' LIMIT 10").df()
```

## Gold: fact_event
`fact_event` should have exactly the clean events.

```python
con.sql("""
    SELECT (SELECT count(*) FROM silver.slv_events) AS slv_events,
           (SELECT count(*) FROM gold.fact_event)   AS fact_event,
           (SELECT count(*) - count(DISTINCT event_key) FROM gold.fact_event) AS duplicate_keys
""").df()
```

```python
con.sql("SELECT * FROM gold.fact_event LIMIT 10").df()
```

Every event points to the item version that was valid at the event time (should be 0).

```python
con.sql("""
    SELECT count(*) AS events_outside_their_version
    FROM gold.fact_event f JOIN gold.dim_item d USING (item_key)
    WHERE NOT (f.event_time >= d.valid_from AND f.event_time < d.valid_to)
""").df()
```

The funnel of day 1: views, add-to-carts, purchases.

```python
con.sql("""
    SELECT t.funnel_step, t.event_type, count(*) AS events, count(DISTINCT f.visitor_id) AS visitors
    FROM gold.fact_event f JOIN gold.dim_event_type t USING (event_type_key)
    GROUP BY 1, 2 ORDER BY 1
""").df()
```
